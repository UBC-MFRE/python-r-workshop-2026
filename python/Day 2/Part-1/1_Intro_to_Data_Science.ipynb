### How to read a function signature

The block above is copied from the pandas documentation. It is not code you run. It lists
every option `pd.read_csv` will accept, and the value each option falls back to when you say
nothing.

Read it like this:

- `filepath_or_buffer` has no `=` after it. That means it is **required**, and it is the only
  required argument. Everything else is optional.
- The lone `*` on the second line is a marker, not an argument. Every argument listed after it
  must be passed **by name**. So `pd.read_csv("f.csv", sep=";")` is legal and
  `pd.read_csv("f.csv", ";")` is not; the second raises
  `TypeError: read_csv() takes 1 positional argument but 2 were given`.
- `header='infer'` shows you the fallback. If you never mention `header`, pandas infers that
  the first line of the file holds the column names.

You will use about five of these forty-odd arguments in a normal week:

| Argument | What it does | When you need it |
| --- | --- | --- |
| `filepath_or_buffer` | where the file is | always |
| `sep` | the character between fields | the file is tab- or semicolon-separated |
| `index_col` | which column becomes the row labels | you already know the key column |
| `na_values` | extra strings to treat as missing | the file writes missing as `-999` or `.` |
| `usecols` | read only some columns | the file is very wide |

Do not memorise the rest. Look them up on the day a file refuses to load correctly.


## How Does `pd.read_csv()` Work in Python?

The function `pd.read_csv()` from the `pandas` library is used to load data from a CSV file into a DataFrame in Python.

A **DataFrame** is a table-like data structure that makes it easy to organize and analyze data.

### How to Use

1. First, import the pandas library:
    ```python
    import pandas as pd
    ```

2. Then, use `pd.read_csv()` with the file path as a string:
    ```python
    data = pd.read_csv("filename.csv")
    ```

This will read the contents of `filename.csv` and store it as a DataFrame called `data`.

### What a CSV file physically is

A CSV file is a plain text file. Nothing more. You could open `data/regional_emissions.csv` in
Notepad and read every character in it. "CSV" stands for comma-separated values: one line per
row, commas between the fields on a line, and the very first line holding the column names.

The real first four lines of `data/regional_emissions.csv` are:

```
Region,Shortnam,1990,2005,2021
Newfoundland and Labrador,NL,9.4,10.2,8.3
Prince Edward Island,PEI,1.8,1.9,1.6
Nova Scotia,NS,19.3,22.8,14.6
```

That is the example the cell above was about to show you.

`pd.read_csv` reads that text, splits each line on the commas, guesses a data type for each
column, and hands back a DataFrame. It never modifies the file. Reading is always safe.

Three decisions it makes for you, all of which matter later:

1. **Column names come from line 1, and they are strings.** The column called `1990` is the
   four-character string `"1990"`, not the number 1990. You have to write
   `regional_emissions["1990"]` with the quotes. Leaving them off gives `KeyError: 1990`.
2. **Some text is silently converted to missing.** `pd.read_csv` treats `n/a`, `NA`, `NULL`,
   `NaN`, and an empty field as missing by default. The last data line of this file has `n/a`
   for Nunavut's 1990 value, so that one cell arrives in Python as `NaN`.
3. **Types are guessed per column.** A column of `9.4, 1.8, 19.3` becomes decimal numbers; a
   column of `NL, PEI, NS` becomes text.

Notice also that the second column is spelled `Shortnam`, not `Shortname`. That is a typo in
the source file, and it is staying. Real data has typos in its headers, and you must type them
exactly as they are or pandas will not find the column.

### The exact error, and three ways to fix it

The error you just saw is:

```
SyntaxError: (unicode error) 'unicodeescape' codec can't decode bytes in position 2-3:
truncated \UXXXXXXXX escape
```

`SyntaxError` means Python could not even read the line, let alone run it. Nothing executed.
The path never reached pandas at all. That is different from every other error in this
notebook, which happen while code is running.

Inside a Python string a backslash is an instruction, not a character:

- `"\n"` is one character, a newline.
- `"\t"` is one character, a tab.
- `"\U"` starts an eight-hex-digit Unicode escape. `"\Users"` hands it `sers` and it fails.

Three fixes, all of which work:

| Fix | Looks like | Notes |
| --- | --- | --- |
| forward slashes | `"C:/Users/name/data/x.csv"` | Windows accepts these too. Simplest. |

| doubled backslashes | `"C:\\Users\\name\\data\\x.csv"` | each `\\` means one literal backslash |

On the professor's aside about which slash is which: a **forward slash** `/` leans forward,
with its top falling to the right. A **backslash** `\` leans the other way. Python is happy
with forward slashes on every operating system, including Windows, which is the easiest thing
to remember.

None of those three fixes the second problem. An absolute path still names one machine. The
next cell fixes both at once.

## 1.	How many rows and columns are in the sector emissions dataset?

When you're working with a Pandas DataFrame object, it's important to know the shape; this is a `(n, q)` tuple where `n` is the length and `q` is the width.

Typically, a dataset's rows are samples while its columns are variables, so `n` tells you the sample size, and `q` tells you the number of variables counted.

### `.shape` in detail

`.shape` is an **attribute**, not a method. You write `df.shape` with no parentheses. Writing
`df.shape()` raises `TypeError: 'tuple' object is not callable`, which is one of the most
common beginner errors in pandas.

The rule of thumb: things that report a stored fact take no parentheses (`.shape`, `.columns`,
`.dtypes`, `.index`, `.values`); things that compute something do (`.sum()`, `.head()`,
`.info()`, `.max()`).

What comes back is a **tuple**: an ordered, fixed-length collection written with round
brackets, like `(32, 8)`. You index it like a list, so `df.shape[0]` is the row count and
`df.shape[1]` is the column count.

Alternatives for the same facts:

| You want | Options | Notes |
| --- | --- | --- |
| row count | `df.shape[0]`, `len(df)` | `len(df)` is shorter and reads better |
| column count | `df.shape[1]`, `len(df.columns)` | `len(df)` gives rows, not columns |
| both at once | `df.shape` | one call, no ambiguity |

Neither the index nor the header line is counted.

because the first line is the header, which becomes column names rather than a row.


Length comes first, then Width. The former is how many rows, i.e. samples the dataframe has; the latter is how many columns, i.e. variables.

### The two pandas data structures, against what you already know

You have already met Python lists and dictionaries. Here is exactly how the pandas types
differ from them, because this is the concept the rest of the notebook stands on.

A **list** is a bare sequence. Positions only, no names.

```python
prices = [9.4, 1.8, 19.3]
prices[0]        # 9.4
```

A **dictionary** maps names to values, but there is no notion of a column of numbers you can
add up in one move.

```python
prices = {"NL": 9.4, "PEI": 1.8, "NS": 19.3}
prices["NL"]     # 9.4
```

A **Series** is both at once: a column of values, each carrying a label, where arithmetic
applies to the whole thing in one go.

```python
prices = pd.Series([9.4, 1.8, 19.3], index=["NL", "PEI", "NS"])
prices["NL"]     # 9.4
prices * 2       # every value doubled, labels kept
prices.sum()     # 30.5
```

A **DataFrame** is a set of Series that share one index. That shared index is what keeps the
columns lined up with each other.

The practical consequences, which the next four cells demonstrate:

- `df["Agriculture"]` returns a **Series**. Its `.shape` is `(32,)`: one number and a trailing
  comma, because a Series has length but no width. The comma is Python's way of writing a
  one-element tuple; without it, `(32)` would just be the number 32.
- `df[["Agriculture"]]` returns a **DataFrame**. A one-column table. Its `.shape` is `(32, 1)`.
- Applying arithmetic to a whole Series at once is called **vectorization**, and it is the
  reason pandas exists. `prices * 2` never writes a loop.

The gotcha: single and double brackets look nearly identical on screen and return different
types. Most functions accept either, so you can go a long time without noticing, and then hit
an `AttributeError` whose message never mentions brackets.


````python
sector_emissions["Agriculture"].shape
````

The rule to memorise: **the brackets decide the type, not the number of columns.**

- `df["col"]` gives a Series.
- `df[["col"]]` gives a one-column DataFrame.

### Slicing: the rule that trips up everyone

Slicing means taking a run of items out of a sequence with `[start:stop]`.

Two rules, and the second one is the one people get wrong:

1. **Counting starts at zero.** The first item sits at position 0, the second at position 1.
2. **The `stop` position is excluded.** `[0:4]` gives you positions 0, 1, 2, and 3. Four items,
   stopping *before* position 4.

Why exclude the stop? Because it makes the arithmetic clean. The number of items you get back
is always `stop - start`. And `[0:4]` followed by `[4:8]` covers everything exactly once, with
no overlap and no gap. If the stop were included you would be writing `+1` and `-1` all day.

Worked example on the eight column names:

```
position:     0          1           2          3           4              5             6                7
name:       Year   Oil and gas  Transport  Buildings  Electricity  Heavy industry  Agriculture  Waste and others
```

| Expression | Result | Why |
| --- | --- | --- |
| `[:4]` | `Year`, `Oil and gas`, `Transport`, `Buildings` | empty start means 0, stops before 4 |
| `[1:]` | everything except `Year` | starts at 1, empty stop means "to the end" |
| `[-3:]` | `Heavy industry`, `Agriculture`, `Waste and others` | negative counts back from the end, so -1 is the last item |
| `[2:5]` | `Transport`, `Buildings`, `Electricity` | three items, which is 5 minus 2 |
| `[::2]` | positions 0, 2, 4, 6 | a third number is the step |
| `[::-1]` | the whole list reversed | a step of -1 |

A slice never errors on being too greedy. `[:100]` on an eight-item list quietly returns all
eight rather than raising. A single index like `[100]` does raise `IndexError`. That difference
catches people.

The gotcha to file away: this exclusive-stop rule holds for lists, tuples, strings, and
`.iloc[]`. It does **not** hold for `.loc[]`, which includes the endpoint. Section 5 returns to
that.

### Missing values, `NaN`, and the boolean-counting trick

`NaN` is pandas' marker for "no value here". It is not zero and it is not an empty string; it
is the absence of data. Two consequences follow, and both bite people:

- Arithmetic involving `NaN` produces `NaN`. `0.6 - NaN` is `NaN`.
- Comparisons involving `NaN` are always `False`, including `NaN == NaN`. This is why you
  cannot test for missing values with `== NaN` and must use `.isna()`.

`.isna()` asks "is this cell missing?" of every cell at once and returns a DataFrame of the
same shape filled with `True` and `False`.

`.isna().sum()` then counts them, and it works because of a Python fact worth knowing on its
own: **`True` counts as 1 and `False` counts as 0.** `True + True + False` evaluates to `2`. So
summing a column of booleans counts the `True` values in it. That idiom appears everywhere in
pandas, not just here.

`.isna().sum()` is the standard one-line missing-data audit. Learn it now and run it on
everything.

Alternatives and relatives:

- `.isnull()` is the same function under a second name. Identical behavior, pick one.
- `.notna()` is the opposite, so `.notna().sum()` counts the values you do have.
- `.isna().sum().sum()` collapses further to a single number: total missing cells in the table.
- `.dropna()` deletes rows containing missing values and `.fillna(0)` replaces them. Neither is
  automatic and neither is always right. Decide deliberately, and write down why.


````python
sector_emissions.isna().sum()
````

Every count is zero. `sector_emissions` has no missing data at all.

Run the same call on the other file and `regional_emissions.isna().sum()` returns `1` for the
`1990` column and `0` for the other four. One line, and you know exactly where the hole is.


````python
sector_emissions.isna()
````

When you're importing a dataset, you probably have some idea of what size it is - roughly how many variables it has, and what ballpark the sample size is in. ### The nested loop, and the `%%time` magic

`%%time` on the very first line of a cell is a **cell magic**: an instruction to Jupyter, not
Python code. It times the whole cell and prints CPU time and wall time underneath the output.
It has to be the first line. Even a comment above it breaks it, which is why the explanatory
comments in these timed cells sit below the magic rather than above it.

The loop itself is two loops, one nested inside the other:

```python
for column in list(sector_emissions.columns)[1:]:   # outer: the 7 sector names
    total = 0                                       # reset, once per column
    for num in sector_emissions[column]:            # inner: the 32 values in that column
        total += num                                # accumulate
    print(column + " total equals: " + str(total))  # report, once per column
```

- The outer loop runs 7 times and the inner runs 32 times for each of those, so 224 additions.
- `total = 0` sits **inside** the outer loop. Move it above and every column's total would pile
  onto the previous one, and the last number printed would be the grand total. That is the
  classic nested-loop bug and it produces plausible-looking wrong answers.
- `total += num` is shorthand for `total = total + num`. Identical meaning, less typing.
- `str(total)` converts the number to text. Python will not join a string to a number with `+`;
  without the conversion you get
  `TypeError: can only concatenate str (not "float") to str`.
- An f-string, `print(f"{column} total equals: {total}")`, avoids the conversion entirely and is
  the modern way to write this line.

Indentation is doing real work here. The body of a `for` loop in Python is defined by
indentation alone, not by braces. The `print` is indented once, so it belongs to the outer loop
and runs 7 times. Indent it twice and it would run 224 times.


````python
%%time

for column in list(sector_emissions.columns)[1:]:
    total = 0
    for num in sector_emissions[column]:
        total += num
    print(column + " total equals: " + str(total))
````

Seven totals, one per sector. `Oil and gas` gives `5214.0`, matching the
single-column loop earlier.

Now look at `Transport total equals: 4745.299999999999`. The true answer is 4745.3. This is not
a pandas bug and not an error in the data. Computers store decimals in binary, and 0.1 has no
exact binary representation for the same reason 1/3 has no exact decimal representation.
Adding 32 such numbers accumulates a rounding error in the fifteenth digit.

You do not fix this by worrying about it. You fix it at the point of display, with
`round(total, 1)` or `f"{total:.1f}"`. And you never test two floats for exact equality; test
`abs(a - b) < 1e-9` instead. This is not a pandas rule, it is true in every programming
language that uses binary floating point, which is all of them.


### Careful: `sector_emissions[1:]` is not `list(sector_emissions.columns)[1:]`

These two expressions look alike and mean different things.

- On a **list of column names**, `[1:]` drops the first *name*, leaving the seven sector
  columns. That is the cell a few above this one.
- On a **DataFrame**, a bare slice like `[1:]` selects *rows*. So `sector_emissions[1:]` is all
  32 rows minus the first one, giving 1991 through 2021, with all eight columns still attached.

You can read the consequence straight off the output below. `Oil and gas` comes back as
`5113.5` rather than `5214.0`, and the difference is exactly `100.5`, the 1990 value that got
sliced away. And a `Year` entry appears reading `62186.0`, which is the sum of the integers
1991 through 2021 and means nothing at all.

This is a good example of a silent wrong answer. Nothing raises. You simply get a total that is
one row short, with a nonsense number sitting next to it.

If what you want is "total every sector column over every year", any of these does it
correctly:

| Expression | How it selects |
| --- | --- |
| `sector_emissions.iloc[:, 1:].sum()` | all rows, columns from position 1 onward |
| `sector_emissions.drop(columns="Year").sum()` | drop the column by name |
| `sector_emissions.set_index("Year").sum()` | move `Year` out of the columns entirely |

All three return `Oil and gas 5214.0`, `Transport 4745.3`, `Buildings 2657.3`,
`Electricity 3085.3`, `Heavy industry 2830.3`, `Agriculture 1948.7`,
`Waste and others 1598.8`.

The general rule: **do not use bare `df[...]` for slicing.** Use `df["col"]` for one column,
`df[["a", "b"]]` for several, and `.loc[]` or `.iloc[]` for anything involving rows. Bare
slicing is the one place where the same brackets change meaning depending on what you put
inside them.

### `axis`: the single most confusing argument in pandas

Almost every pandas aggregation takes an `axis` argument. It answers one question: which
direction do I collapse?

- `axis=0` is the **default**. It moves down the rows and produces one number per column.
- `axis=1` moves across the columns and produces one number per row.

The mnemonic that actually sticks: **`axis` names the dimension that disappears.** With
`axis=0` the rows disappear and you are left with a value per column. With `axis=1` the columns
disappear and you are left with a value per row.

Applied to `regional_emissions[["1990", "2005", "2021"]]`, which is 13 rows by 3 columns:

| Call | Values out | Meaning |
| --- | --- | --- |
| `.sum()` or `.sum(axis=0)` | 3 | the national total for each year |
| `.sum(axis=1)` | 13 | each region's total across the three years |

The cell above is `axis=0`. It returns `1990: 588.6`, `2005: 732.2`, `2021: 670.4`. Those are
Canada-wide emissions in each of the three years, and they are the economically meaningful
numbers here: national emissions rose about 24 percent from 1990 to 2005, then fell about 8
percent by 2021, ending roughly 14 percent above the 1990 level.

One catch worth stating plainly. `588.6` is a sum over **12** regions, not 13, because
Nunavut's 1990 value is `NaN` and `.sum()` skips it. The 1990 and 2005 totals are therefore not
computed over quite the same set of regions. On this data the gap is negligible, since
Nunavut's 2005 figure is 0.6. On other data it will not be.

You can also write `axis="index"` for 0 and `axis="columns"` for 1. Longer, and far harder to
misread six months later. Worth the extra characters in anything someone else will maintain.

`axis=1` now. Thirteen values, one per row, in the same order as the regions.

Row 8 reads `657.5`, which is Alberta: `165.5 + 235.9 + 256.1`. Row 12 reads `1.2`, which is

Is this number meaningful? Barely. It adds emissions from three separate years, which is not a
quantity anyone reports. Summing across a row is genuinely useful when the columns are

The index of a Pandas DataFrame is the set of labels running down the left-hand side; for `sector_emissions`, this means the 0, 1, 2, 3, etc, up to 31 in the bottom row. The index is vital for accessing, aligning, and joining datasets.

For example, if you have two different yearly datasets, one with values of interest rates, and one with unemployment rates, you could use the yearly index to join both together so that all the years match.

We'll use it here to find certain values, with the `.idxmax()`, `.idxmin()`, `.max()`, and `.min()` functions.

But first, we'll need to explicitly set our index as the `"Year"` column, with `.set_index()`

### What an index actually is

Every DataFrame has an index. You never opt in. If you do not choose one, pandas builds a
`RangeIndex` of 0, 1, 2, and so on, which is what you have been looking at down the left edge
of every printout so far.

An index is a set of **labels for rows**. It is not a column. `sector_emissions.shape` reports
`(32, 8)` and the index is not one of those 8. After `set_index("Year")` the shape becomes
`(32, 7)`: the column moved out of the count and into the labels.

Concretely, with `Year` as the index you can write `sector_emissions_index.loc[2015]` and get
the 2015 row by name. With the default index you would have to know that 2015 sits at position
25, and count.

Three things the index buys you:

1. **Labels.** `.idxmax()` returns the index label of the largest value. With `Year` as the
   index that is a year rather than a meaningless row number. That is the entire point of this
   section.
2. **Selection.** `.loc[]` looks rows up by index label, which is section 5.
3. **Alignment.** When you combine two Series, pandas matches them by index label, not by
   position. Add a 1990-to-2021 series to a 2000-to-2021 series and the years line up
   correctly, with the unmatched years coming back as `NaN`. Automatic alignment is the single
   biggest thing pandas does for you, and it is also the source of surprise `NaN` values when
   the labels do not match the way you assumed they would.

A good index is unique and stable. `Year` here is both. `Region` in the other file is both.
Something like "row number in the original spreadsheet" is neither, and makes a bad index.

Three ways to get one:

- `df.set_index("Year")` returns a new DataFrame with `Year` moved out of the columns.
- `pd.read_csv(path, index_col="Year")` does it at load time, in one step, with no extra line.
- `df.reset_index()` undoes it.

### `inplace=True`, and why most pandas code now avoids it

Two ways to say the same thing:

```python
df = df.set_index("Year")             # reassign
df.set_index("Year", inplace=True)    # modify in place
```

The first computes a new DataFrame and points the name `df` at it. The second modifies the
existing DataFrame and returns `None`.

The gotcha that catches everyone exactly once: writing
`df = df.set_index("Year", inplace=True)`. That assigns `None` to `df` and your data is gone.
If you use `inplace`, do not assign. If you assign, do not use `inplace`.

The second gotcha is specific to notebooks. `inplace=True` is **not repeatable**. Run the cell
below twice and the second run raises `KeyError: "None of ['Year'] are in the columns"`,
because after the first run there is no `Year` column left to promote. Reassignment cells
re-run cleanly. `inplace` cells often do not, and in a notebook you will re-run cells
constantly.

The pandas maintainers now discourage `inplace` generally. It rarely saves memory, it cannot be
chained, and it makes your cells order-dependent. Prefer reassignment. The author demonstrates
`inplace` here because you will meet it in other people's code and need to recognise what it
does.


````python
sector_emissions.set_index("Year", inplace = True)
````

### Finding the single largest value in a whole block, with `.stack()`

`.max()` on a DataFrame gives the maximum of each column separately. To get the one largest
number anywhere in a 13 by 3 block, you first have to flatten that block into a single column.
That is what `.stack()` does.

`.stack()` reshapes **wide to long**. Before, one row of the block looks like:

```
                 1990   2005   2021
Nova Scotia      19.3   22.8   14.6
```

After stacking, the same row becomes three entries:

```
Nova Scotia  1990    19.3
             2005    22.8
             2021    14.6
```

Same three numbers, now stacked vertically in one Series, with a two-part label on each value:
region and year. A label made of two parts is called a **MultiIndex**. The 13 by 3 block
becomes a Series of 39 values.

Because the labels are pairs, `.idxmax()` returns a pair. The line

```python
max_region, max_year = stacked.idxmax()
```

is **tuple unpacking**. `.idxmax()` hands back the tuple `('Alberta', '2021')`, and Python
assigns the first element to `max_region` and the second to `max_year` in one step. The number
of names on the left has to match the length of the tuple, or you get
`ValueError: too many values to unpack`.

`future_stack=True` opts into the pandas 3.0 behavior. The older default drops missing values
while stacking, giving 38 entries here instead of 39; with `future_stack=True` the `NaN` for
Nunavut 1990 is kept. It makes no difference to this answer, because `.idxmax()` and
`.idxmin()` skip `NaN` either way, but the newer behavior is the one that will still be
correct next year.

Alternatives for the same question:

- `.max().max()` gives the maximum **value** in two simple steps. Shorter, but it cannot tell
  you which region and year the value came from, which is the whole question here.
- `.melt()` also reshapes wide to long, but returns a plain DataFrame with the labels as
  ordinary columns instead of a MultiIndex. Better when you intend to keep working in long
  form; worse for a one-shot `.idxmax()`.
- `.to_numpy().max()` is the fastest and discards the labels entirely.


````python
# Flatten the 13x3 block into one long Series whose labels are (Region, Year) pairs.
stacked = regional_emissions_index[["1990", "2005", "2021"]].stack(future_stack=True)

max_value = stacked.max()                # the largest number anywhere in the block
max_region, max_year = stacked.idxmax()  # its ("Region", "Year") label, unpacked into two names

print("Highest emissions value:", max_value)
print("Region:", max_region)
print("Year:", max_year)
````

`256.1` megatonnes, Alberta, 2021. Check it against the table above: Alberta's
2021 figure really is the largest single number in the block. The next two largest are
Alberta's own 235.9 in 2005, then Ontario's 203.7 in 2005.

Note the `print` calls. `print("Region:", max_region)` passes **two** arguments, and `print`
joins them with a single space. That is why the output reads `Region: Alberta` with no manual
spacing. Passing `"Region: " + max_region` would also work here but breaks the moment the
second value is a number rather than a string.


````python
min_value = stacked.min()
min_region, min_year = stacked.idxmin()

print("Lowest emissions value:", min_value)
print("Region:", min_region)
print("Year:", min_year)
````

`0.5` megatonnes, Yukon, 1990. The lowest single value in the block.

Nunavut's `NaN` was never a candidate. `.idxmin()` skips missing values rather than treating
them as negative infinity, which is the behavior you want but is worth confirming rather than
assuming.

# 5. Slicing and Selection with `.loc[]` and `.iloc[]`

When we were working with lists, we often would "slice" out specific sections by putting a number, or series of numbers, in behind a list object.

For example:

````python
simple_list = [1, 2, 3, 4, 5]

simple_list[2:4]
````

`simple_list[2:4]` returns `[3, 4]`.

Walk it through. The positions are 0, 1, 2, 3, 4, holding the values 1, 2, 3, 4, 5. Start at
position 2, which holds `3`. Stop **before** position 4. So you get positions 2 and 3, meaning
the values `3` and `4`. Two items, which is `4 - 2`.

If you expected `[3, 4, 5]`, you have just met the exclusive-stop rule the hard way, and the
slicing table earlier in this notebook is worth a second read.


Filtering a DataFrame involves reducing the full dataframe to a subset, where the set of rows match one or more conditions. The `.loc[]` and`.iloc[]` functions can both do this, but approach it in a different way:

- `.loc[]` is based on column/row labels, or names. You name your columns and what condition must apply to them. The dataset returned will be limited to rows where your conditions are met.

- `.iloc[]` does the same, but specifying rows/columns by their integer places.


I'd advise you to prioritize `.loc[]`, because it isn't sensitive to columns having their ordering changed.

### A precise statement of the difference

Both `.loc[]` and `.iloc[]` take two things inside square brackets, separated by a comma: rows
first, columns second.

- `df.loc[rows, columns]`, where rows and columns are given by **label**.
- `df.iloc[rows, columns]`, where rows and columns are given by **integer position**.

Each of the two slots accepts four kinds of thing:

| Kind | `.loc[]` example | `.iloc[]` example |
| --- | --- | --- |
| a single item | `"Nova Scotia"` | `2` |
| a list | `["1990", "2021"]` | `[0, 2]` |
| a slice | `"1990":"2021"` | `1:4` |
| everything | `:` | `:` |

Two differences that will bite you:

1. **Slice endpoints.** `.iloc[1:4]` gives positions 1, 2, 3 and excludes 4, like every other
   Python slice. `.loc["a":"c"]` gives `a`, `b`, **and** `c`, including the endpoint. There is
   a good reason, covered below.
2. **Boolean masks.** `.loc[]` accepts a True/False Series in the row slot, so
   `df.loc[df["Year"] > 2000, "Year"]` works. `.iloc[]` does not; it wants integers, and
   handing it a boolean Series raises
   `NotImplementedError: iLocation based boolean indexing on an integer type is not available`.
   There is a commented-out example of exactly that failure later in this notebook.

Square brackets, never round. `.loc(...)` with parentheses produces a confusing error about
axis names rather than anything that mentions indexing. These are indexers, not functions.


### Data Selection: One Value

Note: `df` is a generic name for any Pandas DataFrame object.

- `df.loc[row_label, column_label]`

- `df.iloc[row_position, column_position]`

````python
regional_emissions[::]
````

`[::]` is a slice with start, stop, and step all left empty, so it means
"everything". It returns the whole table, identical to typing `regional_emissions` on its own.

It is here to demonstrate that DataFrames accept slice syntax at all. In real code, write the
plain name. The one place this form earns its keep is `[::-1]`, which reverses the row order.

Trace the positions yourself. Rows count from 0: `0` is NL, `1` is PEI, `2` is NS. Columns count
from 0 as well, and the index does not participate: `0` Region, `1` Shortnam, `2` "1990",
`3` "2005", `4` "2021". So `[2, 3]` is Nova Scotia's 2005 value, `22.8`. The comment in the cell

The most common mistake here is forgetting that `Region` occupies position 0. If you think of

The same value, `22.8`, reached a different way. Here `2` is a **row label**,

The two agree only because `regional_emissions` currently carries a `RangeIndex`, where the

`regional_emissions.sort_values("2021").loc[2, "2005"]` and you still get Nova Scotia's `22.8`,

Note the quotes around `"2005"`. Column names read from a CSV header are strings even when they
look like numbers. `regional_emissions.loc[2, 2005]` raises `KeyError: 2005`.


To get a row:

- `.iloc[]`: first input the row's index number, and then the index numbers of the columns you want

- `.loc[]`: first input the row's index label, and then the index labels of the columns you want

You can select just one row or column, in which case you just input the number/name. Or, you can input a slice statement for the ones you want (around a `:`), or a list, or just a `:` to return everything.

### 1.1 One row, selecting columns

````python
regional_emissions_index.iloc[1, 1:4] # a slice from a row
````

`iloc[1, 1:4]` gives one row across three columns, so the result is a Series:

Two details in that output are worth stopping on.

- `Name: Prince Edward Island` at the bottom. When you slice a single row out of a DataFrame,
  the row's label becomes the Series' name and the **column names become the index**. The row

- `dtype: object`, even though all three values are decimals. Pulling a row out of a DataFrame
  that contains a text column, here `Shortnam`, forces the result into the one type that can

  element's type as it goes, and a column of numbers sitting in an `object` Series is a
  well-known source of silent bugs.

Compare that with `iloc[:, 1]` a few cells down, which slices a **column** and comes back as

### 2. `.loc[]` examples

### 2.1 One value

### 2.2 One row, selecting columns

````python
regional_emissions_index.loc["Nova Scotia", "1990":"2021"] # a slice from a row
````

### The endpoint rule, stated once and clearly

`"1990":"2021"` returned **three** columns: `1990`, `2005`, and `2021`. The endpoint is
**included**.

Compare that with `.iloc[1:4]`, which returns positions 1, 2, 3 and excludes 4.

| Indexer | Slice written | What you get |
| --- | --- | --- |
| plain list | `[1:4]` | positions 1, 2, 3 |
| `.iloc[]` | `1:4` | positions 1, 2, 3 |
| `.loc[]` | `"1990":"2021"` | all three columns, endpoint included |

The reason for the inconsistency is practical. With positions you can always compute the stop
you want, because you know what number comes next. With labels you usually cannot, because you
do not know which label follows `"2021"`. Asking for "1990 through 2021" and getting everything
up to but excluding 2021 would be useless.

Two conditions on label slices:

- The labels must be **unique**. Duplicated labels make the range ambiguous and pandas raises
  `KeyError`.
- If the index is not sorted, both endpoints have to be present in it exactly, or you get
  `KeyError`. On a sorted index you can slice with endpoints that do not appear at all, and
  pandas takes everything in between.

The result here is again `dtype: object`, for the reason given earlier: this is a row
cross-section of a frame that contains the text column `Shortnam`.


````python
regional_emissions_index.loc["Nova Scotia", ["1990", "2005", "2021"]] # a list from a row (equivalent to the slice!)
````

A list gets the same three columns. With only three year columns present, the
list and the slice are interchangeable.

On a table with 30 year columns the slice would be far shorter to write. And on a table where
you want 1990 and 2021 but not 2005, only the list can express it. Neither form is better in
general; they answer different questions.


````python
regional_emissions_index.loc["Nova Scotia", :] # a whole row
````

The whole row again, `Shortnam` included, via `:` in the column slot.

`regional_emissions_index.loc["Nova Scotia"]` with no comma is equivalent, for the same reason
`.iloc[1]` was equivalent earlier.

All rows, the `"1990"` column. Identical to

The longer `.loc[:, "1990"]` form earns its keep in one situation: **assignment**.

## `.loc[]` versus `.iloc[]`: Which to use and why?

For general selection of columns, I'd recommend you to stick with `.loc[]`. Column names are much more likely to remain stable than data positioning.


#### Say that column names are changed, affecting `.loc[]`, or positions change, affecting `.iloc[]`. Which is worse?

In my opinion, the latter is **much worse**.

- If column names change, `.loc[]` will throw an error message and break.

- But if positions change, `.iloc[]` will grab whatever's there and keep going without batting an eyelid. This could feed in the wrong data at a later point without you having any idea what's going on!

### In Defense of `.iloc[]`: sorting

One functionality `.iloc[]` has which `.loc[]` definitely doesn't relates to sorting. If you use another function to sort a DataFrame by its values in one column, you can use `.iloc[]` to grab the top X, or bottom X, rows, getting you the rows with the highest or lowest values of that variable.

For example, below we sort by 2021 carbon emissions and take the 5 rows at the top of the DataFrame. Sorting least to greatest gives the 5 regions with the lowest emissions in 2021; sorting greatest to least gives the 5 highest.

### `.sort_values()` in detail

`df.sort_values("2021", ascending=True)` returns a new DataFrame with the rows reordered by the
`"2021"` column, smallest first. The arguments worth knowing:

| Argument | Default | Effect |
| --- | --- | --- |
| `by` | required | column name, or a list of them for tie-breaking |
| `ascending` | `True` | `False` for largest first |
| `na_position` | `"last"` | where rows with `NaN` go; `"first"` moves them to the top |
| `inplace` | `False` | same caveats as `set_index` |

Sorting reorders rows but **keeps their index labels attached to them**. That is why the output
below carries row labels 12, 10, 11, 1, 0 rather than 0 through 4. Those are still the original
row numbers, now out of order.

`.iloc[:5]` then takes the first five rows **by position**, which is precisely the author's
point: after a sort, position means "rank" and label means "identity". `.iloc[]` is the tool
that can talk about rank.

Two shorter alternatives:

- `.head(5)` does the same job as `.iloc[:5]` and reads better.
- `regional_emissions.nsmallest(5, "2021")` does the sort and the trim in one call, and
  `.nlargest(5, "2021")` is its opposite.

Prefer `nlargest` and `nsmallest` when you want a top-N and nothing else. Prefer
sort-then-slice when you also want to look at the full ordering.


````python
# get 5 rows with lowest 2021 values

regional_emissions.sort_values("2021", ascending = True).iloc[:5]
````

The five lowest 2021 emitters: Nunavut `0.6`, Yukon `0.7`, Northwest
Territories `1.3`, Prince Edward Island `1.6`, Newfoundland and Labrador `8.3`. All three
territories, plus the two smallest Atlantic provinces.

Nunavut's `NaN` in the `1990` column caused no trouble here, because the sort was on `"2021"`
and that value is present. Had you sorted on `"1990"`, Nunavut would have landed at the bottom
regardless of its value, since `na_position` defaults to `"last"`.


````python
# get 5 rows with highest 2021 values

regional_emissions.sort_values("2021", ascending = False).iloc[:5]
````

And the five highest: Alberta `256.1`, Ontario `150.6`, Quebec `77.5`,
Saskatchewan `67.1`, British Columbia `59.4`.

Those five sum to `610.7` of the national `670.4`, about 91 percent of Canadian emissions from
five of thirteen jurisdictions. Alberta alone is 38 percent.

Only `ascending` changed between the two cells. Everything else is identical.


# Data Selection: Columns

Simply grabbing one column, or a set of them, is easier than grabbing specific bits of data.

For one column, use the format `df["column"]`. Think of this as telling Python to give you just `"column"` from the dataframe called `df`

For multiple columns, you pass a list of columns, like `df[["column_1", "column_2"]]`

### Why `df["col"]` and not `df.col`

pandas also lets you write `regional_emissions.Region` as attribute access, with a dot instead
of brackets. It works. Do not use it, for three reasons:

- It fails on any name containing a space. `sector_emissions.Oil and gas` is not valid Python
  at all.
- It fails on names that start with a digit. `regional_emissions.1990` is a syntax error.
- It silently loses to real DataFrame attributes. A column named `count`, `min`, `shape`, or
  `index` is shadowed by the method or attribute of that name, so `df.count` hands you the
  method rather than your data, and the failure appears somewhere later.

`df["col"]` has none of these problems and always means one thing. Note that both files in this
notebook would break attribute access: `sector_emissions` has `Oil and gas`, `Heavy industry`,
and `Waste and others`, while `regional_emissions` has `1990`, `2005`, and `2021`. The point is
not academic.

The bracket form also lets you build the name at runtime, as in `df[some_variable]`, which the
dot form cannot do at all.


````python
regional_emissions["Region"]
````

One column, single brackets, so a Series: 13 region names carrying their row
labels 0 through 12.


````python
regional_emissions[["1990", "2005", "2021"]]
````

Three columns, a list inside the brackets, so a DataFrame: 13 rows by 3
columns. `Region` and `Shortnam` were left behind.

This is the same operation as `.loc[:, ["1990", "2005", "2021"]]`, written shorter. For plain

````python
# Answer
# The inner expression builds a True/False value for each of the 32 rows; the outer
# brackets keep the True ones. 22 rows survive: 2000 through 2021 inclusive.
sector_emissions_2000s = sector_emissions[sector_emissions["Year"] >= 2000]
sector_emissions_2000s.shape
````

`sector_emissions[sector_emissions["Year"] >= 2000]` is a boolean filter,

`(22, 8)` confirms it worked: 22 years from 2000 through 2021 inclusive, with all 8 columns

````python
# .drop(columns="Year") removes the Year column so its values are not totalled.
# The first .sum() gives 7 sector totals; the second collapses those 7 into 1 number.
canada_total_2000_onwards = sector_emissions_2000s.drop(columns="Year").sum().sum()
canada_total_2000_onwards
````

### Two `.sum()` calls in a row

`sector_emissions_2000s.drop(columns="Year").sum().sum()` is three steps, read left to right:

1. `.drop(columns="Year")` returns a copy without the `Year` column. Without this step the
   total would include the sum of 22 year numbers, which is `44231`, swamping the real answer
   by a factor of nearly three. Note that `columns=` has to be spelled out: `.drop("Year")`
   alone tries to drop a **row** labeled `Year` and raises
   `KeyError: "['Year'] not found in axis"`.
2. The first `.sum()` collapses down the rows, giving 7 numbers, one per sector.
3. The second `.sum()` collapses those 7 into 1.

The result is `15732.699999999999` megatonnes of CO2 equivalent from all sectors, 2000 through
2021. The trailing digits are the binary floating-point artefact from earlier. The honest way
to report it is `round(canada_total_2000_onwards, 1)`, giving `15732.7`.

Alternatives:

- `.to_numpy().sum()` after the drop. Faster, and it sidesteps the two-step aggregation.
- `.stack().sum()`, which flattens to one Series first.
- `.sum(numeric_only=True).sum()` skips the drop by telling pandas to ignore non-numeric

# 6. Filtering


If you're like me, you're probably not paying much attention anymore. The dude up at the front is just going on and on about "locks" and "iLocks".

(Some weird new Apple release?)

But this part is important. A **very** common operation in data cleaning is *"Filtering"*, where you take a large dataset and reduce it to just the samples that fit certain characteristics.


Examples:

- If you've got data on crop returns in Canadian provinces, you might filter to just the samples from Alberta and British Columbia.
- If you've got macroeconomic data, you might filter to just samples from 2000 onwards.
- If you've got a medicinal study, you might filter to just candidates with a certain health condition.

## Can anyone come up with an idea like this - a theoretical dataset, and a criteria you want to filter it by?


### Core idea

There are a few ways to filter data. What we'll be doing revolves around creating a "boolean array", and then selecting rows with it.

A boolean array is created by comparing a logical "True or False" condition with some value in each row. For example, in a dataset on daily weather reports, if the High temperature exceeded 20.0C.

Then, the rest of the expression subsets the dataset to just the rows where this condition is true. We'll work through some examples below.

### A boolean mask, shown concretely

The filtering syntax hides an intermediate object. Run the comparison on its own and you can
see it:

```python
regional_emissions["1990"] <= 10.0
```

produces

```
0      True
1      True
2     False
3     False
4     False
5     False
6     False
7     False
8     False
9     False
10     True
11     True
12    False
Name: 1990, dtype: bool
```

That is the **boolean mask**: one `True` or `False` per row, carrying the same index as the
DataFrame it came from. Nothing has been filtered yet. It is just 13 answers to one question.

Then `regional_emissions[mask]` keeps the rows whose mask entry is `True` and drops the rest.
The matching is done by **index label**, not by position. That is why a mask built from a
different DataFrame will either misalign or raise `IndexingError` rather than quietly lining
up by row number.

Look at row 12, Nunavut. Its `1990` value is `NaN`, and `NaN <= 10.0` evaluates to `False`, not
to `True` and not to missing. **Every comparison against `NaN` is False.** So rows with missing
data are silently dropped by any filter touching that column, whether or not you intended it.
If you want them, test for them separately with `.isna()` and combine the two conditions.


#### 6.1 One condition

The generalized syntax for this goes as such:

`df[df["column"] <boolean operator> <value>]`

- `df` is the DataFrame you're filtering
- `"column"` is the column whose values you're examining
- `<boolean operator>` is an operator, like `<`, `>`, `==`, `<=`, or `>=`
- `<value>` is just some value

What happens then, is that Python uses the `<boolean operator>` term to compare `<value>` with the row's `"column"` value, returning `True` or `False`. Because it does this for each row, it creates our "boolean array"

Finally, `df` is subsetted to just the rows which line up with a `True` value, while the `False` values disappear.

We'll use some examples to get a feel for this.

````python
regional_emissions[regional_emissions["Shortnam"] == "NL"]
````

One row back: Newfoundland and Labrador.

`==` is the comparison operator, "is equal to". `=` is assignment. Writing
`regional_emissions["Shortnam"] = "NL"` would overwrite the entire column with the string
`"NL"` and destroy your data, with no error and no warning. This is the most expensive
one-character typo in the language, and the reason to read filter expressions twice.

String comparison is exact and case sensitive, so `"nl"` matches nothing at all. For looser
matching, `.str.contains("N")` does substring matching and `.isin(["NL", "PEI"])` tests
membership in a list.


What's going on here?

As you can see, we're comparing the `"Shortnam"` column with a string value.

````python
regional_emissions["Shortnam"]
````

The raw column, before any comparison, so you can see what is being tested.
Thirteen two- or three-letter codes. Only position 0 holds `"NL"`, so the mask is `True` exactly
once and the filter returns one row.

To ask for several at once, use `.isin(["NL", "PEI", "NS"])`. Chaining equality tests with the
pipe operator works too, but becomes unreadable past two conditions.


Because only one of these is equal to `"NL"`, we filter down to just that row: Newfoundland and Labrador.

````python
regional_emissions[regional_emissions["1990"] <= 10.0]
````

Four rows survive: Newfoundland and Labrador `9.4`, Prince Edward Island
`1.8`, Yukon `0.5`, Northwest Territories `1.8`. All at or below 10.0 megatonnes in 1990.

The surviving row labels are `0, 1, 10, 11`. Filtering keeps the original labels rather than
renumbering, which is how you can still tell where each row came from. Chain
`.reset_index(drop=True)` if you want a clean 0-to-3 numbering, and only if the original numbers
carry no information.

Nunavut is absent because `NaN <= 10.0` is `False`. Whether that is right for your analysis is a
judgement call, and pandas will not make it for you or tell you it made one.

The operators available: `<`, `<=`, `>`, `>=`, `==`, and `!=` for "not equal". Also
`.between(a, b)`, which is inclusive at both ends, and `.isin([...])` for membership.


Our next example uses a numeric comparison: it makes a boolean array based on if each row's `"1990"` value is equal or less than 10.0, and returns these.

#### Multiple Conditions

Inside of your `df[]` statement, you can include multiple conditions. These can be joined with either the `&` operator (meaning both must be true for a True entry in the boolean array) or an `|` operator (meaning either being true is sufficient).


It's best to wrap the boolean statements inside of round brackets `()` in order to keep them clearly separated and readable; a line break in between is also good practice.


### Why `and` fails and `&` works

The next cell is tagged to fail on purpose. Run it and read the error.

`and` is Python's own logical operator and it works on **one** True or False at a time. Handed
a Series of 13 of them, it has no way to decide what the whole thing means, so it raises:

```
ValueError: The truth value of a Series is ambiguous. Use a.empty, a.bool(), a.item(),
a.any() or a.all().
```

The suggestions in that message are pandas offering you ways to collapse the Series to one
value. None of them is what you want here.

`&` is the elementwise operator. It pairs the two Series up row by row and produces a new Series
of 13 results. That is what filtering needs.

- both conditions: use `&`, never `and`
- either condition: use `|`, the vertical bar, never `or`
- negate a mask: use `~`, never `not`

The parentheses around each condition are **not optional**. In Python `&` binds more tightly
than `<`, so `df["a"] < 10 & df["b"] > 10` is parsed as `df["a"] < (10 & df["b"]) > 10` and
fails with an error that mentions none of this. Always wrap each condition:
`(df["a"] < 10) & (df["b"] > 10)`. The line break the author uses between the two conditions is
purely for readability and is possible only because the whole expression sits inside brackets.


````python
regional_emissions[(regional_emissions["1990"] < 10) and
                   (regional_emissions["2005"] > 10.0)]
````

As you can see, the normal `and` operator throws an error.

````python
regional_emissions[(regional_emissions["1990"] < 10) &
                   (regional_emissions["2005"] > 10.0)]
````

One row: Newfoundland and Labrador, the only region under 10.0 in 1990 and
above 10.0 in 2005. It went from `9.4` to `10.2`.

Check that against the four rows the single-condition filter returned. Of NL, PEI, YT, and NT,
only NL crosses 10.0 by 2005; the other three are still at 1.9, 0.6, and 1.7. Verifying a
filter against a smaller filter you already understand is a good habit, and cheap on a table
this size.


You could also just use two filters in order to get the same result.

````python
regional_emissions_1990_under_10 = regional_emissions[regional_emissions["1990"] < 10]
regional_emissions_1990_under_10_and_2005_above_10 = regional_emissions_1990_under_10[regional_emissions_1990_under_10["2005"] > 10.0]

regional_emissions_1990_under_10_and_2005_above_10
````

The same single row, reached by filtering twice in sequence instead of once
with `&`.

Both are correct. The sequential version is easier to debug, because you can print the
intermediate and see how many rows survived each step. The `&` version is one line and leaves no
half-filtered variable lying around to confuse you later.

A third option sidesteps the naming problem entirely:

```python
regional_emissions.query("`1990` < 10 and `2005` > 10.0")
```

`.query()` takes the whole condition as a string, so `and` is fine inside it and no parentheses
are needed. Column names that are not valid Python identifiers, such as `1990`, need backticks
around them. It is slower on small data and it hides typos until runtime, so treat it as a
matter of taste rather than a recommendation.

# 7. Column Operations

Another very important part of Pandas programming is column operations: you can select columns in the `df["column"]` format, and perform arithmetic calculations with them, likely to create new columns.

### Column arithmetic is elementwise and label-aligned

`regional_emissions["2005"] - regional_emissions["1990"]` does not subtract two numbers. It
subtracts two Series of 13 values each and returns a third Series of 13 values: row 0 minus row
0, row 1 minus row 1, and so on down. No loop and no index variable anywhere. This is
vectorization again, and it is the same machinery that made `.sum()` a one-liner.

The pairing is done on **index label**, not on position. Both operands here come from the same
DataFrame so they match perfectly. When they do not, pandas takes the union of the labels and
fills the gaps with `NaN` rather than raising an error. That is a feature when you meant to
combine two datasets and a trap when you did not, and it is the most common source of
mysterious `NaN` columns.

Assigning to a name that does not exist **creates** a new column:
`df["new"] = <a Series of the right length>`. Assigning to a name that already exists
**overwrites** it with no warning at all. New columns are appended at the right-hand end.

The operators that work this way: `+`, `-`, `*`, `/`, `//`, `%`, `**`, and all the comparisons.
A single number on one side is applied to every row, so `df["col"] * 2` doubles everything.

Alternatives:

- `df.assign(diff=df["2005"] - df["1990"])` returns a **new** DataFrame rather than modifying
  the existing one, which lets you chain and keeps the original clean. Preferred in pipelines.
- `df["2005"].sub(df["1990"], fill_value=0)` is the method form, and it lets you say what a
  missing value should count as instead of propagating `NaN`.
- ``df.eval("diff = `2005` - `1990`")`` is the string form. Rarely worth the loss of clarity.


````python
regional_emissions["1990 to 2005 Difference"] = regional_emissions["2005"] - regional_emissions["1990"]
regional_emissions
````

A sixth column, `1990 to 2005 Difference`, appended on the right. Read a few
values:

- Alberta `70.4`: from `165.5` to `235.9`, by far the largest absolute increase in the country.
- Saskatchewan `25.3` and Ontario `24.6` follow.
- Northwest Territories `-0.1`: the only negative number in the column, and the only
  jurisdiction that emitted less in 2005 than in 1990.
- Nunavut `NaN`, because `0.6 - NaN` is `NaN`. Missing data propagates through arithmetic
  rather than being treated as zero. That is correct behavior: reporting Nunavut as a `0.6`
  increase would be inventing a 1990 observation that does not exist.

Note that the new column name contains spaces. Perfectly legal in pandas, and a concrete
demonstration of why attribute access could never work: `df.1990 to 2005 Difference` is not
something Python could parse.


Here, we subtracted the 1990 values of each province's emissions from the 2005 values, getting the increase or decrease in this time.

````python
regional_emissions["multiplication_example"] = regional_emissions["2005"] * regional_emissions["1990"]

regional_emissions["division_example"] = regional_emissions["2005"] / regional_emissions["1990"]

regional_emissions
````

Two more columns, bringing the frame to 8.

`multiplication_example` is `2005` times `1990`. Alberta's `39041.45` is megatonnes squared,
which is not a unit of anything. The author says as much: it is there to prove the operator
works, not because the number means something.

`division_example` is `2005` divided by `1990`, and this one is meaningful. It is a **ratio**.
Alberta's `1.425378` says 2005 emissions were 1.425 times the 1990 level. Northwest Territories'
`0.944444` says its emissions fell to 94.4 percent of the 1990 level.

Watch the Nunavut row: `NaN` in both new columns, because both operations touch the missing
`1990` value.

One thing to check on data you do not know: division by zero does not raise in pandas the way
it does in plain Python. `1.0 / 0.0` gives `inf` and `0.0 / 0.0` gives `NaN`, both without an
error. There are no zeros in the `1990` column here, but on a denominator you have not
inspected, run `(df["denom"] == 0).sum()` first.


As you can see, we can also do multiplication (`*`) and division (`/`). There's no meaningful interpretation to multiplication, though division can show us the percentage change in this time (if we subtract 1 and multiply by 100).

### The five mistakes that will actually cost you time

1. `df.shape()` instead of `df.shape`. Attributes take no parentheses; methods do.
2. `and` instead of `&` in a filter, and forgetting the parentheses around each condition.
3. `df[1:]` when you meant columns. Bare slicing on a DataFrame selects **rows**.
4. Assuming `.loc[]` excludes its endpoint. It does not. `.iloc[]` does.
5. Forgetting that any comparison against `NaN` is `False`, so filters silently drop rows with
   missing data.

Every one of those produces a plausible answer rather than an error, which is what makes them
expensive. When a number surprises you, check the shape of what you selected before you check
the arithmetic.
